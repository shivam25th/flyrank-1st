## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Decision
Which content pages should the content/SEO team prioritize for refresh or review when their search performance shows signs of decline?

### Who acts on it
The content or SEO team uses the prioritized list to decide which pages deserve human review and possible refresh.

### Cost of a wrong call
If a healthy page is incorrectly prioritized, the team may waste limited time and resources or unnecessarily change content that is already performing well. If a genuinely declining page is missed, the team may lose an opportunity to recover search visibility and traffic.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
df["is_declining_label"] = (
    df["trend_direction"].astype(str).str.lower() == "down"
).astype(int)

print("Total pages:", len(df))
print("Declining pages:", df["is_declining_label"].sum())
print("Declining rate:", round(df["is_declining_label"].mean(), 3))
```

## 3. Quick look at the data (2-3 real numbers)

The dataset contains 30,000 content items across 44 columns. One useful observation is that search volume and observed 90-day impressions are almost uncorrelated in this dataset, with a correlation of approximately 0.001. CTR also varies substantially across search-position tiers, suggesting that search performance cannot be understood from search volume alone.

These observations make content-refresh prioritization worth investigating further, while remaining descriptive rather than causal.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print("Dataset rows:", df.shape[0])
print("Dataset columns:", df.shape[1])

search_impression_corr = df["search_volume"].corr(df["impressions_90d"])
print("Search volume vs impressions_90d correlation:",
      round(search_impression_corr, 3))

visible = df[df["impressions_90d"] >= 100]

ctr_by_position = (
    visible.groupby("position_tier")["ctr"]
    .mean()
    .sort_values(ascending=False)
)

print("\nMean CTR by position tier:")
print(ctr_by_position.round(4).to_string())

```
